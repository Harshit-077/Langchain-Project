## Integrating Langchain + Gemini

#### Switched to Groq because of no credit

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model = 'llama-3.3-70b-versatile',
    temperature = 0.9,
)

response = llm.invoke('Write a paragraph about live on mars in year 2100.')
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.chains import create_retrieval_chain

prompt = PromptTemplate.from_template("You are a content creator. Write me a tweet about {topic}")

chain = prompt | llm
topic = "Artificial intelligence"
result = chain.invoke({"topic": topic})
print(result.content)
```

## System Prompt and Streaming

```python
from langchain_core.messages import HumanMessage, SystemMessage
output = llm.invoke(
    [
        SystemMessage("Answer only YES or NO in German."),
        HumanMessage("is cat a domestic animal?"),
    ]
)
print(output.content)
```

### Streaming

```python
prompt = "Write a paragraph about the mathematical foundation of our universe."
response = llm.invoke(prompt)
print(response.content)
```

```python
for chunk in llm.stream(prompt):
    print(chunk.content, end = '')
```
