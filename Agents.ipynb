```python
from langchain_experimental.utilities import PythonREPL
python_repl = PythonREPL()
python_repl.run('print([n for n in range(1,100) if n%2 == 0])')
```

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv, find_dotenv
from langchain_experimental.tools.python.tool import PythonREPLTool
from langchain_experimental.agents.agent_toolkits import create_python_agent

load_dotenv(find_dotenv(), override=True)

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature = 0,
)

agent_executor = create_python_agent(llm = llm,
                                     tool=PythonREPLTool(),
                                     verbose = True)
agent_executor.invoke('Calculate the square root of factorial of 12 and display it with 4 decimal points')
```

```python
response = agent_executor.invoke('what is the answer of 5.1**7.3?')
response
```
